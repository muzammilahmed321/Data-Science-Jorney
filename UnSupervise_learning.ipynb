```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore', category=UserWarning)
```

```python
X, y_true = make_blobs(
    n_samples=500, #500 data points
    centers=3,
    cluster_std=0.60,
    random_state=42
)
```

```python
X
```

```python
df=pd.DataFrame(X,columns=['Featur1','Featur2'])
```

```python
df
```

```python
scaler=StandardScaler()
x_scaled=scaler.fit_transform(df)
```

```python
import warnings
warnings.filterwarnings('ignore', category=UserWarning)

# 1. Define k_range
k_range = range(1, 11)

# 2. Reset the inertia list explicitly
inertia = []

# 3. Run the loop
for k in k_range:
    kmeans = KMeans(n_clusters=k, n_init=10, random_state=42)
    kmeans.fit(x_scaled)
    inertia.append(kmeans.inertia_)

# 4. Plot (both shapes now match at (10,))
plt.plot(k_range, inertia, marker='o')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method')
plt.show()
```

```python
inertia
```

```python
kmeans_final= KMeans(n_clusters=3,random_state=42)
```

```python
cluster_labels=kmeans_final.fit_predict(x_scaled)
```

```python
cluster_labels
```

```python
df['cluster']=cluster_labels
```

```python
sns.scatterplot(
    x=df['Featur1'],
    y=df['Featur2'],
    hue=df['cluster'],
    palette='viridis'
)
```

```python
from sklearn.cluster import KMeans,DBSCAN
from sklearn.datasets import make_moons
```

```python
X,y_true=make_moons(n_samples=500,noise=0.05,random_state=42)
```

```python
df1=pd.DataFrame(X,columns=["feture1","feature2"])
df1
```

```python
df.head()
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(df)
```

```python
kmeans=KMeans(n_clusters=2,random_state=42)
kmean_labels=kmeans.fit_predict(X_scaled)
```

```python
df1['kmeans_cluster']=kmean_labels
```

```python
sns.scatterplot(x=df1['feture1'],y=df1['feature2'],hue=df1['kmeans_cluster'],palette='tab10')
```

```python
dbscan=DBSCAN(eps=0.3,min_samples=5)
dbscan_labels=dbscan.fit_predict(X_scaled)
df1['dbscan_cluster']=dbscan_labels
```

```python
sns.scatterplot(x=df1['feture1'],y=df1['feature2'],hue=df1['dbscan_cluster'],palette='tab10')
```
