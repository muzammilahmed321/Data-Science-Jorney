```python
#heatmap
penguins.head()
```

```python
coloums=["bill_length_mm","bill_depth_mm","flipper_length_mm","body_mass_g"]
penguins[coloums].corr()
#correlation basicly give relation bettween two coloums
```

```python
sns.heatmap(data=penguins[coloums].corr(),annot=True,cmap="Blues",vmin="-0.2",linewidth="2",linecolor="black")
```

```python
#rug plot
sns.rugplot(
    data=penguins,
    y="body_mass_g",
    hue="species",
    palette="Set2",
    height=0.2
)
```

```python
sns.pairplot(
    data=penguins,
    hue="sex",
    palette="Set2",
    diag_kind="kde"
)
```

```python
sns.pairplot(
    data=penguins,
    hue="sex",
    palette="Set2",
    diag_kind="hist"
)
```

```python
#pair grid-> subplot grid for plotting parirwise reletionship in data set
graph = sns.PairGrid(
    data=penguins,
    hue="sex",
    palette="Set2"
)

graph.map_upper(sns.scatterplot) #upper triangle
graph.map_lower(sns.kdeplot)   #lower triangle
graph.map_diag(sns.histplot)   #digonal

graph.add_legend()


```
