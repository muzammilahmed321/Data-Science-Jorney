```python
#import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("sales_data_sample.csv",encoding="latin1")
print(df)
```

```python
print(df.isnull().sum())
```

```python
df.dropna(subset=['ADDRESSLINE2','STATE','POSTALCODE','TERRITORY'])
print(df)
#data is cleaned
print(df.isnull().sum())
```

```python
type_counts=df['STATUS'].value_counts()
plt.figure(figsize=(6,4))
plt.bar(type_counts.index,type_counts.values,color=['skyblue','orange'])
plt.title('Number of product shipped or in progress')
plt.xlabel('STATUS')
plt.ylabel('Count')
plt.tight_layout()
plt.savefig('product shipped or inprogress.png')
plt.show()

ORDERNUMBER_counts=df['ORDERNUMBER'].value_counts().head(10)
plt.figure(figsize=(8,4))
plt.bar(ORDERNUMBER_counts.index,ORDERNUMBER_counts.values)
plt.title('Top 10 most salled products')
plt.xlabel('ORDERNUMBER')
plt.ylabel('count')
plt.tight_layout()
plt.savefig('Top 10 most salled products.png')
plt.show()
```
