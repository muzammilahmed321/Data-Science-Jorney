```python
#plt.plot(x,y,color='color name',linestyle='line_style',linewidth=value,marker='marker symbol',label='label name')
#plt.xlabel('text')
#plt.xlim() and plt.ylim() it control the values of x axis and y axis
import matplotlib.pyplot as plt
month=[1,2,3,4]
sales=[1000,1500,1200,1700]
plt.plot(month,sales,color='blue',linestyle='--',linewidth=1,marker='o',label='2025 sales data')
plt.xlabel('Months')
plt.ylabel('Sales per month')
plt.title('Monthly Sales Data Report')
plt.legend(loc='lower right')
#optional fuction of legend are upper left:top left corner
#lower right:bottom right corner
#plt.grid()
#optional functionalities
plt.grid(color='gray',linestyle=':',linewidth=1)
plt.xlim(1,4)
plt.ylim(0,2000)
plt.xticks([1,2,3,4],['M1','M2','M3','M4'])
plt.show()
```

Ploting Bar Charts ,Pie Charts and Histograms

```python
#plt.bar(x,height,color='color name',label='name')
import matplotlib.pyplot as plt
product=['A','B','C','D']
sales=[1000,1500,800,1200]
plt.bar(product,sales,color='yellow',label='sakes 2025')
plt.xlabel('Product')
plt.ylabel('Sales')
plt.title('product sales comparision')
plt.legend()
plt.show()
```

```python
#plt.bar(x,height,color='color name',label='name')
import matplotlib.pyplot as plt
product=['A','B','C','D']
sales=[1000,1500,800,1200]
plt.barh(product,sales,color='yellow',label='sakes 2025')
plt.xlabel('Product')
plt.ylabel('Sales')
plt.title('product sales comparision')
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
#plt.pie(values,label=label_list,color=color_list,autopct='%1.1f%%')
#autopct tell about how much persentage each label utilize
#1.1f mean 1 number before decimal 1 number after decimal
#%% used to print % sign
region=['North','South','East','West']
revenue=[3000,2000,1500,1000]
plt.pie(revenue,labels=region,autopct='1%.1f%%',colors=['skyblue','lightgreen','yellow','coral'])
plt.title('Revenue genrated by each region of Pakistan')
plt.legend()
plt.show()
```

```python
#plt.hist(data,bins=number_of_bins,color='colorname',edgecolor='colorname')
import matplotlib.pyplot as plt
score=[45,67,80,70,45,58,67,73,90,84,78,92,73,80,58,77,82,90]
plt.hist(score,bins=5,color='purple',edgecolor='black')
plt.xlabel('Score Range')
plt.ylabel('Number of Students')
plt.title('score distribution of students')
plt.show()
```

Scatter plots

```python
#plt.scatter(x,y,color='color name',marker='marker style',label='label name')
import matplotlib.pyplot as plt
hour_studies=[1,2,3,4,5,6,7,8]
exam_scores=[50,55,60,65,70,75,80,85]
plt.scatter(hour_studies,exam_scores,color='red',marker='*',label='student Data')
plt.xlabel('Hour studied')
plt.ylabel('exam_scores')
plt.title('Relationship Between Studies Time and Exam Score')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#plt.subplot(nrows,ncols,index)
import matplotlib.pyplot as plt
x=[1,2,3,4]
y=[10,20,15,25]
plt.subplot(1,2,1) #1row, 2column, 1st subplot
plt.plot(x,y)
plt.title('line chart')
plt.subplot(1,2,2)#1row, 2 column, 2nd subplot
plt.bar(x,y)
plt.title('Bar chart')
#plt.tight_lay
```

```python
import matplotlib.pyplot as plt
#fig, ax=plt.subplot(nrows,ncols,figsize=(width,height))
fig, ax =plt.subplots(1,2,figsize=(10,5))
x=[1,2,3,4]
y=[10,20,15,25]
ax[0].plot(x,y,color='blue')
ax[0].set_title('line plot')
ax[1].bar(x,y,color='green')
ax[1].set_title('Bar Chart')
fig.suptitle('Comparision of Line and Bar Charts')
plt.tight_layout()
plt.show()
```

tightlayout() is use to fit all over graphs and figure in given canvas
