```python
import pandas as pd
data={
"Name":['Muzammil','shazeeb','rehan','tafseer','ghazanfar','adil','rehman'],
"Age":[18,19,19,20,21,25,26],
"salary":[10000,20000,23000,30000,24000,50000,150000],
"performance":[85,30,75,40,60,83,50]
}
df=pd.DataFrame(data)
print("sample data")
print(df)
#selecting single colomn
print("Name (Single column return series)")
name=df["Name"]
print(name)
#selecting multiple columns
subset=df[["Age","salary"]]
print("\n sub set with name and salary ")
print(subset)
```

```python
#filtering using one condition
high_salary=df[df['salary']>50000]
print("employes with salary>50000")
print(high_salary)
```

```python
#filtering using multiple condition but and condition
filtered=df[(df['Age']>20) & (df['salary']>30000)]
print("employe with age> 20 and salary> 30000")
print(filtered)
```

```python
#using or condition
filtered=df[(df['Age']>20) | (df['performance']>75)]
print("employe with age> 20 or performamce> 75")
print(filtered)
```

```python
import pandas as pd
data={
"Name":['Muzammil',None,'rehan','tafseer','ghazanfar','adil','rehman'],
"Age":[18,None,19,20,21,25,26],
"salary":[10000,None,23000,30000,24000,50000,150000],
"performance":[85,None,75,40,60,83,50]
}
df=pd.DataFrame(data)
print(df)
df['Age'].fillna(df['Age'].mean(),inplace=True)
df['salary'].fillna(df['salary'].mean(),inplace=True)
df['performance'].fillna(df['performance'].mean(),inplace=True)
print("\n")
print(df)
```

```python
import pandas as pd
data={
"Name":['Muzammil','shazeeb','rehan','tafseer','ghazanfar','adil','rehman'],
"Age":[18,None,19,20,21,25,26],
"salary":[10000,None,23000,30000,24000,50000,150000],
"performance":[85,None,75,40,60,83,50]
}
df=pd.DataFrame(data)
print(df)
#linear,polynomial,time
print("\n")
df.interpolate(method="linear",axis=0,inplace=True)
print(df)
```

Merging and Joining:

pd.merge(df1,df2,on="Colomn_Name",how="type of join")
type of join are 
1 iner
2 outer
3 left
4 right

```python
import pandas as pd
#coustomer dataframe
df_customers=pd.DataFrame({
    'CustomerID':[1,2,3],
    'Name':['Muzammil','Saqib','Adil']
    
})
#orders dataframe
df_orders=pd.DataFrame({
 'CustomerID':[1,2,4],
'OrdersAmount':[250,450,350]
})
#merge
df_merged=pd.merge(df_customers,df_orders,on="CustomerID",how="inner")
print("inner join")
print(df_merged)
                          
                         
```

```python
#outer join
#it will merge all the rows and those rows which doesnot match so it will 
#represent it through nan
df_merged=pd.merge(df_customers,df_orders,on="CustomerID",how="outer")
print("outer join")
print(df_merged)
```

```python
#left join 
#it will all the rows of left data frame
df_merged=pd.merge(df_customers,df_orders,on="CustomerID",how="left")
print("left join")
print(df_merged)
```

```python
#right join 
#it will place all the rows of right data frame
df_merged=pd.merge(df_customers,df_orders,on="CustomerID",how="right")
print("right join")
print(df_merged)
```

Concetenations:
Combining data frame 
vertically [row_wise]
horizentally [column_wise]

```python
#pd.concate([df1,df2],axis=0,ignore_index=True)
#ignore index mean ignore the index of new data frame
import pandas as pd
#region 1
df_region1=pd.DataFrame({
    'CoustomerID':[1,2],
    'Name':['Muzammil','Saqib']
})
#region 2
df_region2=pd.DataFrame({
    'CoustomerID':[3,4],
    'Name':['andy','carlo'] 
})
#concatenate vertically
df_concat=pd.concat([df_region1,df_region2],ignore_index=True)
print(df_concat)
```

```python
#concatenate horizently
df_concat=pd.concat([df_region1,df_region2],axis=1,ignore_index=True)
print(df_concat)
```
