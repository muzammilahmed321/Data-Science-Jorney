```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("students(1).csv",encoding="latin1")
print(df)
```

```python
print(df.isnull().sum())
```

```python
df.columns = df.columns.str.strip()
df = df.dropna(subset=[
    'First_Name', 'Last_Name', 'Email', 'Gender', 'Age', 'Department',
    'Midterm_Score', 'Final_Score', 'Assignments_Avg', 'Quizzes_Avg',
    'Parent_Education_Level', 'Stress_Level (1-10)', 'Sleep_Hours_per_Night'
])

```

```python
print(df.isnull().sum())
```

```python
gender_counts=df['Gender'].value_counts()
plt.figure(figsize=(8,6))
plt.pie(gender_counts,labels=gender_counts.index,autopct='%.1f%%')
plt.title('Persentage of students with respect to gender ')
plt.tight_layout()
plt.savefig('gender of diffrent student in university')
plt.show()
```

```python
student_counts=df['Department'].value_counts()
plt.figure(figsize=(6,4))
plt.bar(student_counts.index,student_counts.values)
plt.title('Number of student in each department')
plt.xlabel('Departements')
plt.ylabel('Students')
plt.tight_layout()
plt.savefig('movies_vs_tvShows.png')
plt.show()
```

```python
df.sort_values(by='Age',ascending=True,inplace=True)
print(df)
```

```python
avg_age=df['Age'].mean()
print(avg_age)
```

```python
male_avg = df[df['Gender'] == 'Male']['Final_Score'].mean()
female_avg = df[df['Gender'] == 'Female']['Final_Score'].mean()
plt.bar(['Male', 'Female'], [male_avg, female_avg], color=['blue','pink'])
plt.title('Final average score with respect to gender')
plt.ylabel('Final Score')
plt.show()
```

```python
student_counts=df['Family_Income_Level'].value_counts()
plt.figure(figsize=(6,4))
plt.bar(student_counts.index,student_counts.values)
plt.title('Number of student belong to each family inome level')
plt.xlabel('income levels')
plt.ylabel('Students')
plt.tight_layout()
plt.savefig('incomes_vs_students.png')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['Attendance (%)'], bins=10, color='skyblue', edgecolor='black')
plt.title('Distribution of Attendance Percentage')
plt.xlabel('Attendance (%)')
plt.ylabel('Number of Students')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
student_counts=df['Extracurricular_Activities'].value_counts()
plt.figure(figsize=(8,5))
plt.bar(student_counts.index,student_counts.values, color=['lightgreen','skyblue'])
plt.title('Number of Student involve in extra caricular activites')
plt.xlabel('Extra caricular activities')
plt.ylabel('Number of Students')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['Sleep_Hours_per_Night'], bins=10, color='lightgreen', edgecolor='black')
plt.title('Distribution of sleep hours vs number of student')
plt.xlabel('slepp hours')
plt.ylabel('Number of Students')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
